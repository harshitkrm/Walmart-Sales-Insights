```python
# =========================
# Walmart Sales Analysis
# Exploratory Data Analysis
# =========================

# 1. Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# -------------------------
# 2. Load Dataset
# -------------------------
# Make sure walmart_sales.csv is in the same folder
df = pd.read_csv("walmart_sales.csv")

print("First 5 rows:")
print(df.head())

# -------------------------
# 3. Basic Data Inspection
# -------------------------
print("\nDataset Info:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

# -------------------------
# 4. Feature Engineering
# -------------------------
# Convert Date and Time
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time']).dt.time

# Weekday vs Weekend
df['Day_Type'] = df['Date'].dt.dayofweek.apply(
    lambda x: 'Weekend' if x >= 5 else 'Weekday'
)

# Time of Day Buckets
def time_of_day(t):
    if t.hour < 12:
        return 'Morning'
    elif t.hour < 18:
        return 'Afternoon'
    else:
        return 'Evening'

df['Time_of_Day'] = df['Time'].apply(time_of_day)

print("\nData after feature engineering:")
print(df.head())

# -------------------------
# 5. Branch-wise Total Sales
# -------------------------
branch_sales = df.groupby('Branch')['Total'].sum().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(data=branch_sales, x='Branch', y='Total')
plt.title("Total Sales by Branch")
plt.ylabel("Total Revenue")
plt.xlabel("Branch")
plt.show()

# -------------------------
# 6. Average Transaction Value by Branch
# -------------------------
avg_bill = df.groupby('Branch')['Total'].mean().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(data=avg_bill, x='Branch', y='Total')
plt.title("Average Transaction Value by Branch")
plt.ylabel("Average Bill Value")
plt.xlabel("Branch")
plt.show()

# -------------------------
# 7. Weekday vs Weekend Sales
# -------------------------
day_sales = df.groupby('Day_Type')['Total'].sum().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(data=day_sales, x='Day_Type', y='Total')
plt.title("Sales: Weekday vs Weekend")
plt.ylabel("Total Revenue")
plt.xlabel("Day Type")
plt.show()

# -------------------------
# 8. Time of Day Analysis
# -------------------------
time_sales = df.groupby('Time_of_Day')['Total'].sum().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(
    data=time_sales,
    x='Time_of_Day',
    y='Total',
    order=['Morning', 'Afternoon', 'Evening']
)
plt.title("Sales by Time of Day")
plt.ylabel("Total Revenue")
plt.xlabel("Time of Day")
plt.show()

# -------------------------
# 9. Product Category Contribution
# -------------------------
product_sales = df.groupby('Product line')['Total'].sum().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(data=product_sales, x='Product line', y='Total')
plt.xticks(rotation=45)
plt.title("Revenue by Product Category")
plt.ylabel("Total Revenue")
plt.xlabel("Product Category")
plt.show()

# -------------------------
# 10. Final Observations (printed)
# -------------------------
print("\nKey Observations:")
print("- Sales vary significantly across branches.")
print("- Weekends generally contribute higher revenue than weekdays.")
print("- Afternoon time slots show higher sales compared to mornings and evenings.")
print("- Certain product categories dominate total revenue.")

```
